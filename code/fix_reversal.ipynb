```python
import pandas as pd
import ast
```

```python
## Update nba.csv

file = "../data/raw/nba.csv"
df = pd.read_csv(file)

df["Sequence"] = df["Sequence"].apply(lambda x: str(ast.literal_eval(x)[::-1]))

df = df[['League', 'Season', 'Team', 'Sequence']]

df.to_csv("../data/raw/nba.csv")
```

```python
## Update us_leagues.csv

file_2 = "../data/processed/us_leagues.csv"

df = pd.read_csv(file_2)

nba = df[df['League'] == "NBA"].copy()

nba["Sequence"] = nba["Sequence"].apply(lambda x: str(ast.literal_eval(x)[::-1]))

df.loc[df['League'] == "NBA", "Sequence"] = nba["Sequence"]

df = df[['League', 'Season', 'Team', 'Sequence']]

df
```

```python
df.to_csv(file_2, index=False)
```
