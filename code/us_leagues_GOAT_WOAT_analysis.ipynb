```python
import pandas as pd
from scipy.special import comb
import ast
```

```python
def upperTailProb(n, k):
    prob = 0
    for i in range(k, n + 1):
        choose = comb(n, i, exact=True)
        prob += choose * (0.5 ** n)
    return prob

def lowerTailProb(n, k):
    prob = 0
    for i in range(0, k + 1):
        choose = comb(n, i, exact=True)
        prob += choose * (0.5 ** n)
    return prob

def goatScore(n, k):
    prob = upperTailProb(n, k)
    return 1 / prob if prob > 0 else float('inf')

def woatScore(n, k):
    prob = lowerTailProb(n, k)
    return 1 / prob if prob > 0 else float('inf')
```

```python
df = pd.read_csv("../data/processed/mlb&nba.csv")

df['Sequence_Numeric'] = df['Sequence'].apply(ast.literal_eval)
df['Wins'] = df['Sequence_Numeric'].apply(sum)
df['Games_Played'] = df['Sequence_Numeric'].apply(len)
```

```python
df['GOAT_Score'] = df.apply(lambda row: goatScore(row['Games_Played'], row['Wins']), axis = 1)

df['WOAT_Score'] = df.apply(lambda row: woatScore(row['Games_Played'], row['Wins']), axis = 1)
df.head(1)
```

```python
output = "../data/processed/us_leagues_binomial.csv"

out = df[['League', 'Season', 'Team', 'Sequence', 'GOAT_Score', 'WOAT_Score']]

out.to_csv(output)
```
