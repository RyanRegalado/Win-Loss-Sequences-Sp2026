```python
import pandas as pd
import numpy as np
```

```python

df = pd.read_csv("../data/raw/Matches.csv")

#df = pd.read_csv('data/raw/Matches.csv')
```

```python
df.head()
```

```python
df2 = df[['Division', 'MatchDate', 'HomeTeam', 'AwayTeam', 'FTResult']]
df2.head()
```

```python
df2.shape
```

```python
df3 = df2[(df2['Division'] == 'E0') | (df2['Division'] == 'SP1') | (df2['Division'] == 'D1')]

df3.head()
```

```python
df3.shape
```

```python
df4 = df3.copy()
df4.head()
```

```python
df4['MatchDate'] = pd.to_datetime(df4['MatchDate'])
df4.head()
```

```python
df4['Season'] = df4['Season'] = np.where(df4['MatchDate'].dt.month >=8,
                                        df4['MatchDate'].dt.year,
                                        df4['MatchDate'].dt.year -1)
df4
```

```python
home_df = df4[['Division', 'MatchDate', 'HomeTeam', 'FTResult', 'Season']]
home_df
```

```python
def points(result):
    if result == 'H':
        return 3
    elif result == 'D':
        return 1
    else:
        return 0
```

```python
home_df['Points'] = home_df['FTResult'].apply(points)
```

```python
home_df
```

```python
home_df['Team'] = home_df['HomeTeam']
home_df.head()
```

```python
def away_points(result):
    if result == 'A':
        return 3
    elif result == 'D':
        return 1
    else:
        return 0
```

```python
away_df = df4[['Division', 'MatchDate', 'AwayTeam', 'FTResult', 'Season']]
away_df
```

```python
away_df['Points'] = away_df['FTResult'].apply(away_points)
```

```python
away_df
```

```python
away_df['Team'] = away_df['AwayTeam']
away_df.head()
```

```python
combined_df = pd.concat([home_df, away_df], ignore_index = True)
combined_df
```

```python
combined_df = combined_df.sort_values(['Division', 'Team', 'Season', 'MatchDate'])
combined_df.head()
```

```python
combined_df['Team'] = combined_df['Team'].replace("M'gladbach", "MGladbach")
combined_df['Team'] = combined_df['Team'].replace("Nott'm Forest", "Nottm Forest")
```

```python
sorted_df = (combined_df.groupby(['Division','Season','Team']).agg(Sequence = ('Points', list)).reset_index())

sorted_df.rename(columns={'Division': 'League'}, inplace=True)

sorted_df['League'] = sorted_df['League'].replace("E0", "Premier_League")
sorted_df['League'] = sorted_df['League'].replace("D1", "Bundesliga")
sorted_df['League'] = sorted_df['League'].replace("SP1", "La_Liga")
sorted_df
```

```python
#check if season length is good in sequences
sorted_df['seq_len'] = sorted_df['Sequence'].apply(len)
sorted_df[sorted_df['League'] == 'Bundesliga'].groupby('Season')['seq_len'].mean()
```

```python
expected = {
    'Premier_League': 20,
    'Bundesliga': 18,
    'La_Liga': 20  # adjust as needed
}

counts = sorted_df.groupby(['League', 'Season'])['Team'].nunique()

# Filter to only rows that don't match expected
mask = counts.index.get_level_values('League').map(expected) != counts.values
counts[mask]
```

```python
sorted_df = sorted_df[sorted_df['seq_len'] > 1]
```

```python
sorted_df.to_csv('soccer.csv')
```

```python
sorted_df.to_csv('../data/processed/soccer.csv')
```

# Creating a binary csv removing all draws

```python
def winLossOnly(seq):
    result = []
    for x in seq:
        if x==3:
            result.append(1)
        elif x == 0:
            result.append(0)
    return result
```

```python
dfWLO = sorted_df.copy()
```

```python
dfWLO['Sequence']=dfWLO['Sequence'].apply(winLossOnly)
dfWLO.head()
```

```python
dfWLO.to_csv('soccer_binary.csv', index = False)
```
