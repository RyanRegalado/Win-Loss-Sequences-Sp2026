```python
import pandas as pd

import ast

from win_loss_utils import c1, c2, c3, c1_argmax_n2, c2_argmax_n2, c3_argmax_n2, split
```

```python
soccer = pd.read_csv("../../output/csvs/soccer_second_half_cscores.csv")
us = pd.read_csv("../../output/csvs/us_leagues_second_half_cscores.csv")

# Get list equivalent instead of string

soccer['Sequence_Numeric'] = soccer['Sequence'].apply(ast.literal_eval)
us['Sequence_Numeric'] = us['Sequence'].apply(ast.literal_eval)
```

```python
results = [c1_argmax_n2(seq, len(split(seq)[1] )) for seq in us['Sequence_Numeric']]
us['n1*'], us['c1*'] = zip(*results)

results = [c1_argmax_n2(seq, len(split(seq)[1] )) for seq in soccer['Sequence_Numeric']]
soccer['n1*'], soccer['c1*'] = zip(*results)
```

```python
results = [c2_argmax_n2(seq, len(split(seq)[1] )) for seq in us['Sequence_Numeric']]
us['n2*'], us['c2*'] = zip(*results)

results = [c2_argmax_n2(seq, len(split(seq)[1] )) for seq in soccer['Sequence_Numeric']]
soccer['n2*'], soccer['c2*'] = zip(*results)
```

```python
us = us.drop(['Sequence_Numeric'], axis = 1)
soccer = soccer.drop(['Sequence_Numeric'], axis = 1)
```

```python
filename = "../../output/csvs/us_generalized_cscores"

us_output =  "../../output/csvs/us_generalized_cscores.csv"
soccer_output =  "../../output/csvs/soccer_generalized_cscores.csv"

us.to_csv(us_output, index = False)
soccer.to_csv(soccer_output, index = False)
```
