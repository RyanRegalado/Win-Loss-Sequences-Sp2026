```python
import pandas as pd
import numpy as np
from win_loss_utils import c3_argmax_n2_fast
import ast
```

```python
soccer_binary = pd.read_csv("../../data/processed/soccer_binary.csv")
soccer = pd.read_csv("../../data/processed/soccer.csv")
us_leagues = pd.read_csv("../../data/processed/us_leagues.csv")

soccer = soccer[['League', 'Season', 'Team', 'Sequence']]

soccer['Sequence_Numeric'] = soccer['Sequence'].apply(ast.literal_eval)
soccer_binary['Sequence_Numeric'] = soccer_binary['Sequence'].apply(ast.literal_eval)
us_leagues['Sequence_Numeric'] = us_leagues['Sequence'].apply(ast.literal_eval)
```

```python
N = 1000

results = [c3_argmax_n2_fast(np.array(seq), N) for seq in us_leagues['Sequence_Numeric']]
us_leagues['n3_star'], us_leagues['max_val'] = zip(*results)

results = [c3_argmax_n2_fast(np.array(seq), N) for seq in soccer['Sequence_Numeric']]
soccer['n3_star'], soccer['max_val'] = zip(*results)

results = [c3_argmax_n2_fast(np.array(seq), N) for seq in soccer_binary['Sequence_Numeric']]
soccer_binary['n3_star'], soccer_binary['max_val'] = zip(*results)
```

```python
## Recalculate top 100 in max_val column

us_indexes = list(us_leagues.sort_values('max_val', ascending=False).head(100).index)
soccer_indexes = list(soccer.sort_values('max_val', ascending=False).head(100).index)
soccer_binary_indexes = list(soccer_binary.sort_values('max_val', ascending=False).head(100).index)

N = 100000

mask = us_leagues.index.isin(us_indexes)
_results = us_leagues.loc[mask, 'Sequence_Numeric'].apply(lambda seq: c3_argmax_n2_fast(np.array(seq), N))
us_leagues.loc[mask, 'n3_star'] = _results.apply(lambda x: x[0])
us_leagues.loc[mask, 'max_val'] = _results.apply(lambda x: x[1])

mask = soccer.index.isin(soccer_indexes)
_results = soccer.loc[mask, 'Sequence_Numeric'].apply(lambda seq: c3_argmax_n2_fast(np.array(seq), N))
soccer.loc[mask, 'n3_star'] = _results.apply(lambda x: x[0])
soccer.loc[mask, 'max_val'] = _results.apply(lambda x: x[1])

mask = soccer_binary.index.isin(soccer_binary_indexes)
_results = soccer_binary.loc[mask, 'Sequence_Numeric'].apply(lambda seq: c3_argmax_n2_fast(np.array(seq), N))
soccer_binary.loc[mask, 'n3_star'] = _results.apply(lambda x: x[0])
soccer_binary.loc[mask, 'max_val'] = _results.apply(lambda x: x[1])
```
