```python
import pandas as pd
import numpy as np
from win_loss_utils import c3_argmax_n2, c3_old
import ast
```

```python
soccer_binary = pd.read_csv("../../data/processed/soccer_binary.csv")
soccer = pd.read_csv("../../data/processed/soccer.csv")
us_leagues = pd.read_csv("../../data/processed/us_leagues.csv")

soccer = soccer[['League', 'Season', 'Team', 'Sequence']]

soccer['Sequence_Numeric'] = soccer['Sequence'].apply(ast.literal_eval)
soccer_binary['Sequence_Numeric'] = soccer_binary['Sequence'].apply(ast.literal_eval)
us_leagues['Sequence_Numeric'] = us_leagues['Sequence'].apply(ast.literal_eval)
```

```python
us_leagues_test = us_leagues.sample(100)
```

```python
results = us_leagues_test["Sequence_Numeric"].apply(
    lambda seq: c3_argmax_n2(np.array(seq), N=1000)
)

# Split the (n3_star, max_val) tuples into separate columns
us_leagues_test["n3_star"], us_leagues_test["max_val"] = zip(*results)
```

```python
results = us_leagues["Sequence_Numeric"].apply(
    lambda seq: c3_argmax_n2(np.array(seq), N=1000)
)

# Split the (n3_star, max_val) tuples into separate columns
us_leagues["n3_star"], us_leagues["max_val"] = zip(*results)
```

```python
results = soccer["Sequence_Numeric"].apply(
    lambda seq: c3_argmax_n2(np.array(seq), N=1000)
)

# Split the (n3_star, max_val) tuples into separate columns
soccer["n3_star"], soccer["max_val"] = zip(*results)
```

```python
results = soccer_binary["Sequence_Numeric"].apply(
    lambda seq: c3_argmax_n2(np.array(seq), N=1000)
)

# Split the (n3_star, max_val) tuples into separate columns
soccer_binary["n3_star"], soccer_binary["max_val"] = zip(*results)
```
