```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
us = pd.read_csv("csvs/us_generalized_cscores.csv")
soccer = pd.read_csv("csvs/soccer_generalized_cscores.csv")
```

```python
us['log_c1*'] = np.log(us['c1*'])
soccer['log_c1*'] = np.log(soccer['c1*'])

us['log_c2*'] = np.log(us['c2*'])
soccer['log_c2*'] = np.log(soccer['c2*'])
```

```python
us["name"] = us["Season"].astype(str) + " " + us["Team"]
soccer["name"] = soccer["Season"].astype(str) + " " + soccer["Team"]
```

```python
top10_us_c1 = us.nlargest(10, 'log_c1*')

plt.figure(figsize=(10,6))
sns.barplot(x='name', y='log_c1*', data=top10_us_c1, palette='Blues_d')
plt.xlabel('Team')
plt.ylabel('log_c1*')
plt.title('Top 10 US Teams by log_c1*')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
plt.close()
```

```python
top10_us_c2 = us.nlargest(10, 'log_c2*')

plt.figure(figsize=(10,6))
sns.barplot(x='name', y='log_c2*', data=top10_us_c2, palette='Blues_d')
plt.xlabel('Team')
plt.ylabel('log_c2*')
plt.title('Top 10 US Teams by log_c2*')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
plt.close()
```

```python
top10_soccer_c1 = soccer.nlargest(10, 'log_c1*')

plt.figure(figsize=(10,6))
sns.barplot(x='name', y='log_c1*', data=top10_soccer_c1, palette='Blues_d')
plt.xlabel('Team')
plt.ylabel('log_c1*')
plt.title('Top 10 Soccer Teams by log_c1*')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
plt.close()
```

```python
top10_soccer_c2 = soccer.nlargest(10, 'log_c2*')

plt.figure(figsize=(10,6))
sns.barplot(x='name', y='log_c2*', data=top10_soccer_c2, palette='Blues_d')
plt.xlabel('Team')
plt.ylabel('log_c2*')
plt.title('Top 10 Soccer Teams by log_c2*')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
plt.close()
```
