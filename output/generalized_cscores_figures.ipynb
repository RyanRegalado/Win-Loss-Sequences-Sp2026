```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
master = pd.read_csv("csvs/master_database.csv")

master["name"] = master["season"].astype(str) + " " + master["team"]

us = master[master['type'] == 'us']
soccer = master[master['type'] == 'soccer_binary']
soccer_ternary =  master[master['type'] == 'soccer_ternary']
```

```python
us['log_c1*'] = np.log(us['c1*'])
soccer['log_c1*'] = np.log(soccer['c1*'])

us['log_c2*'] = np.log(us['c2*'])
soccer['log_c2*'] = np.log(soccer['c2*'])

us['log_c3*'] = np.log(us['c3*'])
soccer['log_c3*'] = np.log(soccer['c3*'])
soccer_ternary['log_c3*'] = np.log(soccer_ternary['c3*'])
```

```python
top10_us_c1 = us.nlargest(10, 'log_c1*')

plt.figure(figsize=(18,7))
sns.barplot(x='name', y='log_c1*', data=top10_us_c1, palette='tab10', hue='league')
plt.xlabel(None)
plt.ylabel('Logarithmic C1* Collapse Score', fontsize=16)
plt.title('Top 10 US Teams by log_c1*')
plt.legend(fontsize=14, title_fontsize=15)
plt.xticks(fontsize=14)
plt.xticks(rotation=45, fontsize=14, ha='right')
plt.savefig("figures/generalized_top10_us_c1.png", dpi=300, bbox_inches='tight')
plt.show()
plt.close()
```

```python
top10_us_c2 = us.nlargest(10, 'log_c2*')

plt.figure(figsize=(18,7))
sns.barplot(x='name', y='log_c2*', data=top10_us_c2, palette='tab10', hue='league')
plt.xlabel(None)
plt.ylabel('Logarithmic C2* Collapse Score', fontsize=16)
plt.title('Top 10 US Teams by log_c2*')
plt.legend(fontsize=14, title_fontsize=15)
plt.xticks(fontsize=14)
plt.xticks(rotation=45, fontsize=14, ha='right')
plt.savefig("figures/generalized_top10_us_c2.png", dpi=300, bbox_inches='tight')
plt.show()
plt.close()
```

```python
top10_soccer_c2 = soccer.nlargest(10, 'log_c2*')

fig, ax = plt.subplots(figsize=(18, 7))

sns.barplot(
    x='name',
    y='log_c2*',
    data=top10_soccer_c2,
    hue='league',
    palette='tab10',
    ax=ax
)

ax.set_xlabel(None)
ax.set_ylabel('Logarithmic C2* Collapse Score', fontsize=16)
ax.set_title('Top 10 Soccer Teams by log_c2*')
ax.legend(fontsize=14, title_fontsize=15)
ax.tick_params(axis='x', rotation=45, labelsize=14)
```
