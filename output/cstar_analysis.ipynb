```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import math
```

```python
leagues = pd.read_csv('../output/csvs/master_database.csv')
for col in ['c1*', 'c2*', 'c3*']:
    leagues[f'log_{col}'] = np.where(leagues[col] > 0, np.log(leagues[col]), np.nan)

soccer = leagues[leagues['type'] == 'soccer_binary']
PL = soccer[soccer['league'] == 'Premier_League']
```

## Premier League C* metrics

### Log C1* 

```python
PL['log_c1*'].value_counts().reset_index()
```

```python
PL['log_c1*'].describe().reset_index()
```

```python
plt.hist(PL['log_c1*'],color='green', edgecolor='black')
plt.ylabel('Number of times value has appeared')
plt.xlabel('Log C1* Value')
plt.title('Distribution of Log C1* for Premier League')
plt.show()
```
